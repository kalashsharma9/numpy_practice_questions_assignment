### Q32. Create an array from `[5, 10, 15, 20, 25]`, slice out `[10, 15, 20]`, then normalize that slice to range 0–1 using min‑max formula and print it.

```python
import numpy as np


arr = np.array([5, 10, 15, 20, 25])


slice_arr = arr[1:4]


normalized_slice = (slice_arr - slice_arr.min()) / (slice_arr.max() - slice_arr.min())


print(normalized_slice)
```

### Q39. Using NumPy, standardize the `math` scores to **z-scores**: `(x - mean) / std` and print the first 5 standardized values.

```python
import numpy as np


math_array = df['math'].to_numpy()

math_mean = np.mean(math_array)
math_std = np.std(math_array)


standardized_math = (math_array - math_mean) / math_std


print("First 5 standardized math scores:", standardized_math[:5])
```

### Q40. Using NumPy logical operations, count how many students scored **≥ 70 in all three subjects** and print that count.

```python
import numpy as np


math_array = df['math'].to_numpy()
english_array = df['english'].to_numpy()
science_array = df['science'].to_numpy()


students_above_70 = np.logical_and(np.logical_and(math_array >= 70, english_array >= 70), science_array >= 70)


count_students_above_70 = np.sum(students_above_70)


print("Number of students who scored ≥ 70 in all subjects:", count_students_above_70)
```
